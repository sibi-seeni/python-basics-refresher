```python
import pandas as pd
```

```python
def get_data():
  entry = input("Enter age, weight (lbs) & height (cm), separated by space: ").strip()
  age, weight, height = entry.split()
  return int(age), int(weight), int(height)
```

```python
def BMI(W, H):
  bmi = (0.4536 * W) / ((0.01 * H) ** 2)
  if bmi < 18.5:
    bmi_bin = 'underweight'
  elif bmi >= 18.5 and bmi < 25:
    bmi_bin = 'healthy'
  elif bmi >= 25 and bmi < 30:
    bmi_bin = 'overweight'
  else:
    bmi_bin = 'obese'
  return round(bmi, 1), bmi_bin
```

```python
def main():
  print("Program to store data")
  bmi_dict = {}
  loop = 'y'
  age_list = []
  bmi_list = []
  bmi_bin_list = []
  while True:
    # Loop to compute BMI and category
    print("================================================")
    age, weight, height = get_data()
    bmi, bmi_bin = BMI(weight, height)
    print(f'Calculated BMI and category: {bmi}, {bmi_bin}')

    # Add the age, bmi and bin to their lists
    age_list += [age]
    bmi_list += [bmi]
    bmi_bin_list += [bmi_bin]

    # Store it all in a dictionary
    bmi_dict = {'Age': age_list, 'BMI': bmi_list, 'BMI_Bin': bmi_bin_list}

    # Convert dict to dataframe
    df = pd.DataFrame(bmi_dict)
    print("\nThe stored pandas DataFrame:")
    print("----------------------------------")
    print(df)
    print("----------------------------------")

    # To loop or not
    loop = input("Enter again (y or n)? ")
    if loop.lower() != 'y':
      print("================================================")
      print("Okay.Bye!")
      break
```

```python
# Calling main() professionally
if __name__ == '__main__':
  main()
```
