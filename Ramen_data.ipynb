10. In the Country column, replace 'USA' with 'United States'.
    
    replace(): https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.replace.html

```python
df = df.replace('USA', 'United States')
df.head(5)
```

### Questions

1. How many countries are represented in the data?

```python
uniq_countries = df['Country'].nunique()
print(f'Unique countries represented in the data: {uniq_countries}')
```

2. Which three countries have the highest average rating?
    
    groupby(): https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html

```python
average = df.groupby("Country")["Stars"].mean()
average.sort_values(ascending=False).head(3)
```

3. Which three countries have the lowest average rating?

```python
average.sort_values(ascending=True).head(3)
```

4. Which three countries have the most brands, and how many brands does each of these countries have?

```python
most_brands = df.groupby("Country")["Brand"].nunique()
most_brands.sort_values(ascending=False).head(3)
```
